```python
# ==================================================
# ARUNA - INITIAL DATA AUDIT
# ==================================================

import pandas as pd
from pathlib import Path

# 1. Load dataset

CURRENT_DIR = Path.cwd()

# Mendukung dua kondisi:
# 1. Kernel dijalankan dari folder utama ARUNA/
# 2. Kernel dijalankan dari folder ARUNA/notebooks/
if (CURRENT_DIR / "data" / "reference").exists():
    PROJECT_ROOT = CURRENT_DIR
elif (CURRENT_DIR.parent / "data" / "reference").exists():
    PROJECT_ROOT = CURRENT_DIR.parent
else:
    raise FileNotFoundError(
        "Folder proyek ARUNA tidak ditemukan. "
        "Pastikan notebook berada di dalam folder notebooks/ "
        "atau kernel dijalankan dari folder utama ARUNA/."
    )

DATASET_PATH = PROJECT_ROOT / "data" / "reference" / "thesis data set.xlsx"

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset tidak ditemukan: {DATASET_PATH}\n"
        "Pastikan thesis data set.xlsx sudah ditempatkan di "
        "ARUNA/data/reference/."
    )

print("Root proyek:", PROJECT_ROOT)
print("Lokasi dataset:", DATASET_PATH)


df = pd.read_excel(DATASET_PATH)

# 2. Ringkas nama kolom ke-4 dan ke-5
df.rename(columns={
    df.columns[3]: "Agree",
    df.columns[4]: "Disagree"
}, inplace=True)

# 3. Ringkasan dataset
print("=== ARUNA DATA AUDIT ===")
print("Observasi :", df.shape[0])
print("Variabel  :", df.shape[1])
print("Duplikasi :", df.duplicated().sum())
print("Missing   :", df.isnull().sum().sum())

# 4. Informasi variabel
audit = pd.DataFrame({
    "Tipe Data": df.dtypes.astype(str),
    "Missing": df.isnull().sum(),
    "Nilai Unik": df.nunique()
})

display(audit)

# 5. Sampel data
print("\n=== SAMPLE DATA ===")
display(df.head())

print("\nAudit awal selesai. Dataset asli belum diubah.")
```

```python
# ==================================================
# ARUNA - DATA VALIDITY CHECK (FIX)
# ==================================================

# 1. Cari kolom berdasarkan kata kunci
keywords = [
    "undergraduate student",
    "strongly agree",
    "How old are you?",
    "socioeconomic status"
]

for keyword in keywords:
    print(f"\n=== Pencarian: {keyword} ===")

    matching_cols = [
        col for col in df.columns
        if keyword.lower() in col.lower()
    ]

    for col in matching_cols:
        print(f"\nNama kolom: {repr(col)}")
        print(df[col].value_counts(dropna=False))

# 2. Periksa 14 butir MHC-SF
mhc_columns = [
    col for col in df.columns
    if "During the past month" in col
]

print("\nJumlah butir MHC-SF:", len(mhc_columns))

for i, col in enumerate(mhc_columns, 1):
    print(f"\nButir {i}:")
    print(df[col].value_counts(dropna=False))
```

```python
# ==================================================
# ARUNA - MHC-SF RESPONSE QUALITY CHECK
# ==================================================

import pandas as pd

# Salinan untuk analisis, bukan mengubah data mentah
analysis_df = df.copy()

# Identifikasi 14 kolom MHC-SF
mhc_columns = [
    col for col in analysis_df.columns
    if "During the past month" in col
]

# Ubah Prefer not to answer menjadi missing
for col in mhc_columns:
    analysis_df[col] = analysis_df[col].replace(
        "Prefer not to answer", pd.NA
    )

# Hitung missing pada 14 butir
missing_per_item = analysis_df[mhc_columns].isna().sum()

print("=== MISSING MHC-SF ===")
print(missing_per_item)

# Hitung responden dengan minimal satu jawaban kosong
respondents_with_missing = (
    analysis_df[mhc_columns].isna().any(axis=1).sum()
)

print("\nResponden dengan minimal satu respons tidak tersedia:",
      respondents_with_missing)

print("Responden dengan 14 jawaban tersedia:",
      len(analysis_df) - respondents_with_missing)

# Periksa responden yang memilih Other pada status mahasiswa
status_col = [
    col for col in df.columns
    if "undergraduate student" in col.lower()
    and "[Other]" not in col
][0]

print("\n=== STATUS RESPONDEN DI LUAR PILIHAN UTAMA ===")
display(
    df.loc[
        df[status_col].eq("Other"),
        [status_col]
    ]
)
```

```python
# ==================================================
# ARUNA - FINAL SAMPLE CHECK
# ==================================================

# Salinan data untuk analisis
analysis_df = df.copy()

# Identifikasi kolom status mahasiswa
status_col = next(
    col for col in df.columns
    if "undergraduate student" in col.lower()
    and "[Other]" not in col
)

# Keluarkan responden yang bukan mahasiswa aktif
analysis_df = analysis_df[
    analysis_df[status_col] == "Yes"
].copy()

# Identifikasi 14 butir MHC-SF
mhc_columns = [
    col for col in analysis_df.columns
    if "During the past month" in col
]

# Perlakukan Prefer not to answer sebagai missing
analysis_df[mhc_columns] = analysis_df[mhc_columns].replace(
    "Prefer not to answer", pd.NA
)

print("=== HASIL SELEKSI RESPONDEN ===")
print("Jumlah awal:", len(df))
print("Jumlah setelah seleksi:", len(analysis_df))
print("Responden dengan 14 jawaban lengkap:",
      analysis_df[mhc_columns].notna().all(axis=1).sum())
print("Responden dengan minimal satu jawaban tidak tersedia:",
      analysis_df[mhc_columns].isna().any(axis=1).sum())
```

```python
# ==================================================
# ARUNA - MISSING RESPONSE PATTERN
# ==================================================

# Hitung jumlah jawaban tidak tersedia per responden
missing_per_respondent = analysis_df[mhc_columns].isna().sum(axis=1)

print("=== POLA MISSING PER RESPONDEN ===")
print(missing_per_respondent.value_counts().sort_index())

# Persentase missing per butir
print("\n=== MISSING PER BUTIR MHC-SF ===")

missing_summary = pd.DataFrame({
    "Jumlah Missing": analysis_df[mhc_columns].isna().sum(),
    "Persentase Missing (%)": (
        analysis_df[mhc_columns].isna().mean() * 100
    ).round(2)
})

display(missing_summary)

# Missing per dimensi
dimensions = {
    "Emotional": mhc_columns[0:3],
    "Social": mhc_columns[3:8],
    "Psychological": mhc_columns[8:14]
}

print("\n=== MISSING PER DIMENSI ===")

for dimension, columns in dimensions.items():
    total = analysis_df[columns].isna().sum().sum()
    print(f"{dimension}: {total} missing dari {len(analysis_df) * len(columns)} respons")
```
